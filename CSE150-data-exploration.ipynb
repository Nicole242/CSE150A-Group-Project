```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
data = pd.read_csv('bank.csv', delimiter=';', quotechar='"')
```

```python
print("Number of observations:",data.shape[0])
print("\nData information:")
print(data.info())
print("\nThere are",data.duplicated().sum(),"duplicated rows")
```

```python
average_balance = data["balance"].mean()
print("Average balance:",average_balance)
print("\nMaximum balance:",data["balance"].max())
print("\nMinimum balance:",data["balance"].min())
average_age = data["age"].mean()
print("\nAverage age:",average_age)
print("\nMaximum age", data["age"].max())
print("\nMinimum age", data["age"].min())

print("\nEmployment probability:")

for job in data["job"].unique():
    prob=data[data["job"]==job].shape[0]/data.shape[0]
    if(job=="admin."):
      print(job,"\t\t:",prob)
      continue
    print(job,"\t:",prob)

print("\nMarriage probability:")
for marital_status in data["marital"].unique():
    prob=data[data["marital"]==marital_status].shape[0]/data.shape[0]
    if(marital_status=="single"):
      print(marital_status,"\t\t:",prob)
      continue
    print(marital_status,"\t:",prob)

print("\nEducation probability:")
for education in data["education"].unique():
    prob=data[data["education"]==education].shape[0]/data.shape[0]
    print(education,"\t:",prob)
print("\nLoan probability")
for loan in data["loan"].unique():
    prob=data[data["loan"]==loan].shape[0]/data.shape[0]
    print(loan,"\t\t:",prob)
```

```python
data["age"].plot(kind="hist", bins=30, edgecolor="black")
plt.xlabel("Age")
plt.ylabel("Frequency")
plt.title("Age histogram")
plt.show()

data["balance"].plot(kind="hist", bins=30, edgecolor="black")
plt.xlabel("Balance")
plt.ylabel("Frequency")
plt.title("Balance histogram")
plt.show()
```
